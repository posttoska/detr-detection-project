```python
# multi head self att
```

```python
num_heads = 8
d_head = 256//8
```

```python
q = torch.randn(1, 400, 256)
k = torch.randn(1, 400, 256)
v = torch.randn(1, 400, 256)
```

```python
Qb, Ql, Qe = q.shape
Kb, Kl, Ke = k.shape
Vb, Vl, Ve = v.shape
```

```python
broadcast_shape_Q = (Qb, Ql, num_heads, d_head)
broadcast_shape_K = (Kb, Kl, num_heads, d_head)
broadcast_shape_V = (Vb, Vl, num_heads, d_head)
```

```python
q = q.view(broadcast_shape_Q)
k = k.view(broadcast_shape_K)
v = v.view(broadcast_shape_V)
```

```python
print(q.shape)
print(v.shape)
print(k.shape)
```

```python
q = q.transpose(1, 2)
k = k.transpose(1, 2)
v = v.transpose(1, 2)
```

```python
print(q.shape)
print(v.shape)
print(k.shape)
```

```python
qk = q @ k.transpose(2, 3)
qk.shape
```

```python
qk /= math.sqrt(d_head)
qk = F.softmax(qk, dim=-1)
```

```python
out = qk @ v
out.shape
```

```python
out = out.transpose(1, 2)
out.shape
```

```python
out = out.reshape(1, 400, 256)
out.shape
```
