```python
import math
import torch
import torch.nn as nn
from torch.nn import functional as F
```

```python
q = torch.randn(1, 25, 256)
k = torch.randn(1, 400, 256)
v = torch.randn(1, 400, 256)
```

```python
print(q.shape)
print(k.shape)
print(v.shape)
```

```python
Qb, Ql, Qe = q.shape
Kb, Kl, Ke = k.shape
Vb, Vl, Ve = v.shape
```

```python
broadcast_q = (Qb, Ql, 8, Qe//8)
broadcast_k = (Kb, Kl, 8, Ke//8)
broadcast_v = (Vb, Vl, 8, Ve//8)
```

```python
print(broadcast_q)
print(broadcast_k)
print(broadcast_v)
```

```python
q_lin = nn.Linear(256, 256)
k_lin = nn.Linear(256, 256)
v_lin = nn.Linear(256, 256)
```

```python
q = q_lin(q)
k = k_lin(k)
v = k_lin(v)
```

```python
q = q.view(broadcast_q)
k = k.view(broadcast_k)
v = v.view(broadcast_v)
```

```python
print(q.shape)
print(k.shape)
print(v.shape)
```

```python
q = q.transpose(1, 2)
k = k.transpose(1, 2)
v = v.transpose(1, 2)
```

```python
print(q.shape)
print(k.shape)
print(v.shape)
```

```python
qk = q @ k.transpose(2, 3)
```

```python
qk.shape
```

```python
qk /= math.sqrt(32)
qk = F.softmax(qk, dim=-1)
```

```python
att_map = qk.mean(dim=1)
att_map.shape
```
