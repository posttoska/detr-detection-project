```python
import torch
import torch.nn as nn
```

```python
# LAYER 1 conv2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        # First Convolution Layer: 7x7, stride=2, padding=3
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=64, kernel_size=7, stride=2, padding=3)
        
    def forward(self, x):
        x = self.conv1(x)  
        return x

model = ResNetSim()
input_image = torch.randn(1, 3, 640, 640)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 1 pool2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        # Max Pooling Layer: 3x3, stride=2, padding=1
        self.pool = nn.MaxPool2d(kernel_size=3, stride=2, padding=1)
        
    def forward(self, x):
        x = self.pool(x)   # Apply maxpool
        return x

model = ResNetSim()
input_image = torch.randn(1, 64, 320, 320)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 2-7 conv2d2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        # 3) Next conv (conv layer #2): 3x3, stride=1, padding=1, 64→64
        self.conv2 = nn.Conv2d(64, 64, kernel_size=3, stride=1, padding=1, bias=False)
        
    def forward(self, x):
        x = self.conv2(x)            # (B, 64, 160, 160)  -- padding=1 keeps size
        return x

model = ResNetSim()
input_image = torch.randn(1, 64, 160, 160)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 8 conv2d2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        # 3) Next conv (conv layer #2): 3x3, stride=1, padding=1, 64→64
        self.conv8 = nn.Conv2d(64, 128, kernel_size=3, stride=2, padding=1, bias=False)
        
    def forward(self, x):
        x = self.conv8(x)  
        return x

model = ResNetSim()
input_image = torch.randn(1, 64, 160, 160)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 9-15 conv2d2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        # 3) Next conv (conv layer #2): 3x3, stride=1, padding=1, 64→64
        self.conv9  = nn.Conv2d(128, 128, 3, 1, 1, bias=False)
        self.conv10 = nn.Conv2d(128, 128, 3, 1, 1, bias=False)
        self.conv11 = nn.Conv2d(128, 128, 3, 1, 1, bias=False)
        self.conv12 = nn.Conv2d(128, 128, 3, 1, 1, bias=False)
        self.conv13 = nn.Conv2d(128, 128, 3, 1, 1, bias=False)
        self.conv14 = nn.Conv2d(128, 128, 3, 1, 1, bias=False)
        self.conv15 = nn.Conv2d(128, 128, 3, 1, 1, bias=False)
        
    def forward(self, x):
        x = self.conv9(x)
        x = self.conv10(x)
        x = self.conv11(x)
        x = self.conv12(x)
        x = self.conv13(x)
        x = self.conv14(x)
        x = self.conv15(x) 
        return x

model = ResNetSim()
input_image = torch.randn(1, 128, 80, 80)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 16 conv2d2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        self.conv16 = nn.Conv2d(128, 256, 3, 2, 1, bias=False)  # downsample -> (B,256,40,40)
        
    def forward(self, x):
        x = self.conv16(x)   # (B,256,40,40)
        return x

model = ResNetSim()
input_image = torch.randn(1, 128, 80, 80)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 17-27 conv2d2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        # 3) Next conv (conv layer #2): 3x3, stride=1, padding=1, 64→64
        self.conv17 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv18 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv19 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv20 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv21 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv22 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv23 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv24 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv25 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv26 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        self.conv27 = nn.Conv2d(256, 256, 3, 1, 1, bias=False)
        
    def forward(self, x):
        x = self.conv17(x)  
        x = self.conv18(x)
        x = self.conv19(x)
        x = self.conv20(x)
        x = self.conv21(x)
        x = self.conv22(x)
        x = self.conv23(x)
        x = self.conv24(x)
        x = self.conv25(x)
        x = self.conv26(x)
        x = self.conv27(x)
        return x

model = ResNetSim()
input_image = torch.randn(1, 256, 40, 40)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 28 conv2d2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        self.conv28 = nn.Conv2d(256, 512, 3, 2, 1, bias=False)  # -> (B,512,20,20)
        
    def forward(self, x):
        x = self.conv28(x)   # (B,512,20,20)
        return x

model = ResNetSim()
input_image = torch.randn(1, 256, 40, 40)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 29-33 conv2d2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        self.conv29 = nn.Conv2d(512, 512, 3, 1, 1, bias=False)
        self.conv30 = nn.Conv2d(512, 512, 3, 1, 1, bias=False)
        self.conv31 = nn.Conv2d(512, 512, 3, 1, 1, bias=False)
        self.conv32 = nn.Conv2d(512, 512, 3, 1, 1, bias=False)
        self.conv33 = nn.Conv2d(512, 512, 3, 1, 1, bias=False)
        
    def forward(self, x):
        x = self.conv29(x)
        x = self.conv30(x)
        x = self.conv31(x)
        x = self.conv32(x)
        x = self.conv33(x)
        return x

model = ResNetSim()
input_image = torch.randn(1, 512, 20, 20)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 34 AdaptiveAvgPool2d
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        self.avg = nn.AdaptiveAvgPool2d((1,1))
        
        
    def forward(self, x):
        x = self.avg(x)        # (B,512,20,20) -> (B,512,1,1)
        return x

model = ResNetSim()
input_image = torch.randn(1, 512, 20, 20)
output = model(input_image)
print(f"Output shape: {output.shape}")
```

```python
# LAYER 34 MLP
class ResNetSim(nn.Module):
    def __init__(self):
        super(ResNetSim, self).__init__()
        self.fc  = nn.Linear(512, 1000)
        
        
    def forward(self, x):
        x = torch.flatten(x, 1)    # -> (B, 512)
        return self.fc(x)          # -> (B, 1000)

model = ResNetSim()
input_image = torch.randn(1, 512, 1, 1)
output = model(input_image)
print(f"Output shape: {output.shape}")
```
